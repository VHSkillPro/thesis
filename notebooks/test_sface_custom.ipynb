```python
import cv2
from lib.face_detector.yunet import YuNetDetector
from lib.face_recognizer.sface_custom import SFaceCustomRecognizer

yunet = YuNetDetector()
sface_custom = SFaceCustomRecognizer(
    model_path="weights/Backbone_MobileFaceNet_KD_CE.onnx"
)


def get_infer(image_path: str):
    image = cv2.imread(image_path)

    face, scale = yunet.detect_single_multiscale(image)
    if face is None:
        return None
    h, w = image.shape[:2]
    scaled_image = cv2.resize(image, (int(w * scale), int(h * scale)))

    embed = sface_custom.infer(scaled_image, face)
    return embed


print(
    sface_custom.similarity(
        get_infer("datasets/cards/21T1020340_TheSinhVien.jpg"),
        get_infer(
            "datasets/selfies/21T1020340_NgoVanHai/Hai_NgoVan_21T1020340_ChanDung.jpg"
        ),
    )
)
```

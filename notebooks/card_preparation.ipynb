```python
import os, pathlib

THESINHVIEN_DIR = "datasets/cards/TheSinhVien"

for filename in os.listdir(THESINHVIEN_DIR):
    filepath = os.path.join(THESINHVIEN_DIR, filename)

    filenamewithoutext = pathlib.Path(filename).stem
    ext = pathlib.Path(filename).suffix

    os.rename(filepath, filenamewithoutext + "_TheSinhVien" + ext)
```

```python
import os, pathlib

THETHUVIEN_DIR = "datasets/cards/TheThuVien"

for filename in os.listdir(THETHUVIEN_DIR):
    filepath = os.path.join(THETHUVIEN_DIR, filename)

    filenamewithoutext = pathlib.Path(filename).stem
    ext = pathlib.Path(filename).suffix

    os.rename(filepath, THETHUVIEN_DIR + "/" + filenamewithoutext + "_TheThuVien" + ext)
```

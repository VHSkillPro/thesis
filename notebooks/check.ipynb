```python
import os, cv2
import os.path as osp
from tqdm import tqdm
from lib.face_detector.yunet import YuNetDetector

CARDS_DIR = "datasets/_cards"
face_detection = YuNetDetector()

for card_name in tqdm(os.listdir(CARDS_DIR)):
    card_path = osp.join(CARDS_DIR, card_name)
    card = cv2.imread(card_path)

    face, _ = face_detection.detect_single_multiscale(card)
    if face is None:
        print(f"Face not detected in {card_name}")
        continue
```

```python
import os, cv2
import os.path as osp
from tqdm import tqdm
from lib.face_detector.retinaface import RetinaFaceDetector

CARDS_DIR = "datasets/_cards"
face_detection = RetinaFaceDetector()

for card_name in tqdm(os.listdir(CARDS_DIR)):
    card_path = osp.join(CARDS_DIR, card_name)
    card = cv2.imread(card_path)

    faces = face_detection.detect(card)
    if len(faces) == 0:
        print(f"Face not detected in {card_name}")
        continue
    if len(faces) > 1:
        print(f"Multiple faces detected in {card_name}")
        continue
```

```python
import os, cv2
import os.path as osp
from matplotlib import pyplot as plt
from tqdm import tqdm
from lib.face_detector.yunet import YuNetDetector

SELFIES_DIR = "datasets/_selfies"
face_detection = YuNetDetector()

for selfie_name in tqdm(os.listdir(SELFIES_DIR)):
    selfie_images_dir = osp.join(SELFIES_DIR, selfie_name)
    selfie_images_name = os.listdir(selfie_images_dir)

    for selfie_image_name in selfie_images_name:
        selfie_image_path = osp.join(selfie_images_dir, selfie_image_name)
        selfie_image = cv2.imread(selfie_image_path)

        face, scale = face_detection.detect_single_multiscale(selfie_image)
        if face is None:
            print(f"Face not detected in {selfie_name}-{selfie_image_name}")
            continue

        scaled_image = cv2.resize(
            selfie_image,
            (int(selfie_image.shape[1] * scale), int(selfie_image.shape[0] * scale)),
        )
        vis = face_detection.visualize(scaled_image, [face])
        plt.imsave(
            f"temp/{selfie_name}-{selfie_image_name}",
            cv2.cvtColor(vis, cv2.COLOR_BGR2RGB),
        )
```

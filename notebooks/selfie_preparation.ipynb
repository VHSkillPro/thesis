```python
import os, cv2

SELFIES_DIR = "datasets/selfies"

for student_name in os.listdir(SELFIES_DIR):
    student_dir = os.path.join(SELFIES_DIR, student_name)

    for selfie_name in os.listdir(student_dir):
        selfie_path = os.path.join(student_dir, selfie_name)
        selfie = cv2.imread(selfie_path)

        h, w = selfie.shape[:2]
        if h < w:
            selfie = cv2.rotate(selfie, cv2.ROTATE_90_CLOCKWISE)
            cv2.imwrite(selfie_path, selfie)
```
